```python
num = int(input("Enter a number: "))
sum_div = 0
for i in range(1, num):
    if num % i == 0:
      sum_div += i
if sum_div == num and num > 0:
  print(f"{num} is a Perfect Number")
else:
  print(f"{num} is not a Perfect Number")
```

```python
def convert(val, unit):
    if unit == "C":
        return val * 9/5 + 32, val + 273.15
    elif unit == "F":
        c = (val - 32) * 5/9
        return c, c + 273.15
    elif unit == "K":
        c = val - 273.15
        return c, c * 9/5 + 32
print(convert(25, "C"))
```

```python
def check_password(password):
    if len(password) < 8:
        print("Weak Password!")
        return False
    if not any(char.isdigit() for char in password):
        print("Weak Password!")
        return False  
    if not any(char.islower() for char in password):
        print("Weak Password!")
        return False 
    if not any(char.isupper() for char in password):
        print("Weak Password!")
        return False     
    special_chars = '!#@$%^&()_+'
    if not any(char in special_chars for char in password):
        print("Weak Password!")
        return False
    print("Strong Password!")
    return True
user_password = input("Enter your password: ")
check_password(user_password)
```

```python
def calculate_cart():
    total = 0
    while True:
        price = input("Enter item price (or 'done' to stop): ")
        if price.lower() == 'done':
            break
        
        total += float(price)
    return total
print("Total Cost:", calculate_cart())
```

```python
def is_palindrome(s):
    reversed_s = ""
    for char in s:
        reversed_s = char + reversed_s

    if s == reversed_s:
        return True
    else:
        return False


user_input = input("Enter a word: ")

if is_palindrome(user_input):
    print("Yes")
else:
    print("No")
```

```python
def is_palindrome(s):
    print(s.capitalize())                       # capitalize(): Capitalizes ONLY the very first character of the string
    print(s.title())                            # title(): Capitalizes the first letter of EVERY word in the string
    print(s.upper())                            # upper(): Converts ALL characters in the string to uppercase
    print(s.lower())                            # lower(): Converts ALL characters in the string to lowercase
    print(s.replace("a", "&"))                  # replace(): Replaces all occurrences of "a" with "&"
    
    s = s.lower().replace(" ", "")              # Clean the string by converting to lowercase and removing spaces
    return s == s[::-1]                         # Return True if the cleaned string equals its reverse, False otherwise

print(is_palindrome(" a man a plan a canal Pananma"))
```

```python
# WAP to find the factorial using recursion
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)
n=int(input("enter the number whose factorial is need to be find = "))
print(factorial(n))
```

```python
# wap porgress to define a function to sort the elemetns of list is descending order.
def sort_descending(lst):
    n = len(lst)
    for i in range(n):
        for j in range(0, n - i - 1):
            if lst[j] < lst[j + 1]:
                lst[j], lst[j + 1] = lst[j + 1], lst[j]
    return lst
my_list = [15, 3, 8, 42, 11, 7]
print(sort_descending(my_list))
```

```python
# wap to define function to find the given no is prime.
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

num = int(input("Enter a number: "))

if is_prime(num):
    print("Prime")
else:
    print("Not prime")
```
