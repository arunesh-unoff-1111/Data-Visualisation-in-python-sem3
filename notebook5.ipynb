```python
## FUNCTIONS
# Skeleton of functions
# def <function_name>(<parameters>){
#           --------------------------
#           ------------------------------
#    return <o/p>
# }
# the value which is passed temproraly in parameters is called arguments or postional argument
# the default argument is a preset value given to a parameter in a function
# the keyword argument is an argument passed to a function preceded by the parameter's name (name=value).
# arbirary arguments : 2 types - 1-- simple arg -2-- keyword arg
```

```python
# WAP to sum a two numbers using functions
def add_numbers(a, b):
    return a + b

a = float(input("Enter the 1st number = "))
b = float(input("Enter the 2nd number = "))
result = add_numbers(a,b)
print(f"the sum of {a} and {b} is {result}")
```

```python
# WAP to check wheather a number is od or even using functions
def check_odd_even(number):
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"

num = int(input("Enter a number: "))
print(check_odd_even(num))
```

```python
# WAP to calculate the area of circle using functions.
def area_of_circle(r, pi):
    return (r ** 2 * pi)

r = float(input("Enter the radius: "))
pi = 3.14
area = area_of_circle(r, pi)
print(f"The area of the circle is: {area}")
```

```python
def addition(*args):
    print(type(args))
    print(args)
    total_sum = 0
    for i in args:
        total_sum = total_sum + i
    return total_sum

# Example usage:
# result = addition(5, 10, 15)
# print(result)  # Output: 30
```

```python
def addition(**kwargs):
    print(type(kwargs))
    print(kwargs)
    total_sum = 0
    for value in kwargs.values():
        total_sum = total_sum + value
    return total_sum

# Example usage:
result = addition(a=5, b=10, c=15)
print(result)  # Output: 30
```

```python
def addition(**kwargs):
    for i in kwargs.items():
        print(i)
    for i in kwargs.keys():
        print(i)
    for i in kwargs.values():
        print(i)
addition(a=5, b=10, c=15)
```
