```python
import pandas as pd

data = [13, 34, 40, 41, 36]
year = [2002, 2002, 2001, 2010, 2001]
s = pd.Series(data, index=year, name='data')

# 1. Access data for 2002 (returns all rows matching 2002)
print("Data for 2002:")
print(s[2002])

# 2. Check data type of the values in the Series
print("\nData Type:")
print(s.dtype)

# 3. Accessing 2003 safely (since 2003 is NOT in your index)
# Using .get() prevents a KeyError if the year doesn't exist
print("\nData for 2003:")
print(s.get(2003, "Year not found"))
```
