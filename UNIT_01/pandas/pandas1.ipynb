```python
import pandas as pd

# 1. Create the Series
s = pd.Series([10, 20, 30, 40, 50], index=['r1', 'r2', 'r3', 'r4', 'r5'])
print("--- Original Series ---")
print(s)

# 2. Scalar Arithmetic (Broadcasting)
print("\n--- Scalar Broadcasting ---")
print("Addition (+ 5):\n", s + 5)
print("Subtraction (- 10):\n", s - 10)
print("Multiplication (* 2):\n", s * 2)
print("Division (/ 5):\n", s / 5)
print("Exponentiation (** 2):\n", s ** 2)

# 3. Series-to-Series Operations (Index Aligned)
s2 = pd.Series([1, 2, 3, 4, 5], index=['r1', 'r2', 'r3', 'r4', 'r5'])
print("\n--- Series-to-Series Operations ---")
print("Addition (s + s2):\n", s + s2)
print("Subtraction (s - s2):\n", s - s2)
print("Multiplication (s * s2):\n", s * s2)

# 4. Conditional / Boolean Operations (Filtering)
print("\n--- Filtering & Conditions ---")
print("Greater than 30 (s > 30):\n", s > 30)
print("Filtered Data (s[s > 30]):\n", s[s > 30])

# 5. Built-in Statistical Methods
print("\n--- Summary Statistics ---")
print(f"Sum: {s.sum()}")
print(f"Mean: {s.mean()}")
print(f"Max: {s.max()}")
print(f"Min: {s.min()}")
print(f"Standard Deviation: {s.std():.2f}")
```

```python
# When performing arithmetic operations between two pandas Series, pandas automatically aligns the data based on matching index labels rather than index positions. If both Series share the same index labels, their values are computed together, but if an index label exists in one Series and not the other, pandas fills the missing result with `NaN` (Not a Number) and automatically promotes the data type to float.
s1=pd.Series([12,13,14,15])
s2=pd.Series([3,2,3,4,5,6])
print(s1.index)
print(s2.index)
s1+s2
```

### Reset Index 
* The `reset_index()` function is used to generate a new DataFrame or Series with the index reset.
* This is useful when the index needs to be treated as a column, or when the current index needs to be reset to the default integer index for subsequent operations.

`Series.reset_index(self, level-None , drop=False , name=None , inplace=false)`

```python
import pandas as pd

# Setup a MultiIndex Series to fully demonstrate level, drop, and name
index = pd.MultiIndex.from_tuples([
    ('Group1', 'r1'),
    ('Group1', 'r2'),
    ('Group2', 'r3'),
    ('Group2', 'r4')
], names=['Group', 'Row'])

s = pd.Series([10, 20, 30, 40], index=index)
print("--- Original MultiIndex Series ---")
print(s)

# 1. Default reset_index() 
# Converts all index levels into DataFrame columns and gives a default range index.
print("\n--- 1. Default reset_index() ---")
print(s.reset_index())

# 2. drop=True
# Drops the existing index entirely instead of converting it into columns (returns a Series).
print("\n--- 2. drop=True ---")
print(s.reset_index(drop=True))

# 3. name= (Renaming the data column)
# Useful when resetting a single-index Series so the data column gets a custom name.
s_single = pd.Series([10, 20, 30], index=['r1', 'r2', 'r3'])
print("\n--- 3. name='Score' (Single Index) ---")
print(s_single.reset_index(name='Score'))

# 4. level= (Resetting a specific index level)
# If you have a MultiIndex, you can choose to reset only one specific level.
print("\n--- 4. level='Group' ---")
print(s.reset_index(level='Group'))

# 5. inplace=True
# Modifies the Series in-place rather than returning a new object (no output returned).
s_copy = s.copy()
s_copy.reset_index(drop=True, inplace=True)
print("\n--- 5. inplace=True Result ---")
print(s_copy)
```
